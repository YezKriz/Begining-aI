```python
# Student Grading System

mark = int(input("Enter the student's mark: "))

if mark >= 90 and mark <= 100:
    grade = "A"
elif mark >= 80 and mark <= 89:
    grade = "B"
elif mark >= 70 and mark <= 79:
    grade = "C"
elif mark >= 60 and mark <= 69:
    grade = "D"
elif mark < 60 and mark >= 0:
    grade = "E"
else:
    grade = "Invalid mark"

print("Grade:", grade)
```
